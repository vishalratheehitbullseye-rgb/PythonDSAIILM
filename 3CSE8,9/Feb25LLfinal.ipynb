```python
# cycle in a linked list
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, x):
#         self.val = x
#         self.next = None

# LC 141

class Solution:
    def hasCycle(self, head: Optional[ListNode]) -> bool:
        slow,fast = head,head
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
            if slow == fast : return True
        return False
```

```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, x):
#         self.val = x
#         self.next = None

class Solution:
    def detectCycle(self, head: Optional[ListNode]) -> Optional[ListNode]:
        slow, fast = head, head
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
            if slow == fast : 
                break
        if fast == None or fast.next == None :
            return 
        fast = head
        while slow != fast : 
            slow = slow.next
            fast = fast.next
        return slow
```

```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
        if l1 == None : return l2
        if l2 == None : return l1
        ans = None
        curans = ans
        curl1 = l1
        curl2 = l2
        carry = 0
        while curl1 or curl2 or carry>0 :
            summ = carry
            if curl1 : summ += curl1.val
            if curl2 : summ += curl2.val
            if ans == None :
                ans = ListNode(summ%10)
                curans = ans
            else : 
                curans.next = ListNode(summ%10)
                curans = curans.next
            carry = summ//10
            if curl1 : curl1 = curl1.next
            if curl2 : curl2 = curl2.next
        return ans
```
