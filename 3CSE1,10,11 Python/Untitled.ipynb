```python
def f(nums , ind , flag , summ ) :
  #base
  if ind == len(nums) : return summ
  #rec
  if flag == False :
    if nums[ind] == 6 :
      flag = True;
    else : summ += nums[ind]
  else :
    if nums[ind] == 7: flag = False
  return f(nums, ind+1, flag, summ)
  

def sum67(nums):
  
  return f(nums, 0, False, 0)
```
