```python
#factorials
# 1 to n, n to 1

def OneToN (n) :
    # base case
    if n<1 : return
    # rec case
    OneToN(n-1)
    print(n)

def NToOne (n) :
    # base case
    if n<1 : return
    # rec case
    print(n)
    NToOne(n-1)

# OneToN(7)
NToOne(7)

```

```python
# x^y using recursion

def powr (x : int , y : int) :
    # base case
    if y == 0 : return 1
    # rec case
    return powr(x, y-1)*x

print(powr(2,5))
print(2**5)

# Hw --> Write rec code for O(log(y)) powr function
# Also check the time complexity of `**` operator
    
```

```python
# sum elements of a list recursively (assume homogenous lists)
# min/max of a list recursively (assume homogenous lists)

a = [1,2,3,4,5]

def listSummer (a : list[int], ind : int):
    # base case
    if ind == len(a) : return 0
    # rec case
    return a[ind] + listSummer(a, ind+1)

def minElement ( a : list[int] , ind : int ) :
    #base
    if ind == len(a) : return 999999999
    #rec case
    minAhead = minElement(a, ind+1)
    return min(a[ind], minAhead)
    

# print(listSummer(a, 0))
print(minElement(a, 0))

#HW
# check whether a string is palindrome or not (recursively)
# check whether a number is palindrome or not


```
